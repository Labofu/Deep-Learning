```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/Users/sayedmohammadnafis/Desktop/ML CSV file/placement.csv')
```

```python
df.head()
```

```python
df = df.drop('Unnamed: 0', axis=1)
```

```python
df.head()
```

```python
x = df.drop(columns=['placement'])

y = df['placement']
```

```python
from sklearn.model_selection import train_test_split

x_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
from sklearn.linear_model import Perceptron
p = Perceptron()
```

```python
p.fit(x_train, y_train)
```

```python
p.coef_
```

```python
p.intercept_
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(x_train.values, y_train.values, clf=p, legend=2)
```

```python
from sklearn.metrics import accuracy_score

y_pred = p.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Test Accuracy:", accuracy)
```

```python
print("Training Accuracy:", p.score(x_train, y_train))
print("Test Accuracy:", p.score(X_test, y_test))
```
